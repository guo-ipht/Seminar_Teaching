# example t-test

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n_sample = 2
S1 = np.random.normal(0, 1, n_sample)
S2 = np.random.normal(0.1, 1, n_sample)
S3 = np.random.normal(2, 1, n_sample)

fig1 = plt.figure(figsize=(4, 3))

### plot data from the sampling
plt.scatter([1]*n_sample, S1, c='r')
plt.scatter([2]*n_sample, S2, c='g')
plt.scatter([3]*n_sample, S3, c='b')

### mark mean values
plt.plot([0.75, 1.25], [0, 0], c='r')
plt.plot([1.75, 2.25], [0.1, 0.1], c='g')
plt.plot([2.75, 3.25], [2, 2], c='b')

plt.xticks([1,2,3], labels=['S1', 'S2', 'S3'])
plt.title('sample size: ' + str(n_sample))
```

```python
n_sample = 50
S1 = np.random.normal(0, 1, n_sample)
S2 = np.random.normal(0.1, 1, n_sample)
S3 = np.random.normal(2, 1, n_sample)

fig1 = plt.figure(figsize=(4, 3))

### plot data from the sampling
plt.boxplot(S1, positions=[1])
plt.boxplot(S2, positions=[2])
plt.boxplot(S3, positions=[3])

plt.xticks([1,2,3], labels=['S1', 'S2', 'S3'])
plt.title('sample size: ' + str(n_sample))
```

```python
from scipy import stats
```

```python
### two-sample two-sided test (S1 vs S2)
t_stat, p_val = stats.ttest_ind(S1, S2, equal_var=True, alternative='two-sided')  
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```

```python
### two-sample one-sided test (S1 vs S2)
t_stat, p_val = stats.ttest_ind(S1, S2, equal_var=True, alternative='less')  
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```

```python
### two-sample one-sided test (S1 vs S2)
t_stat, p_val = stats.ttest_ind(S1, S2, equal_var=True, alternative='greater')  
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```

```python
t_stat, p_val = stats.fisher_exact(table, alternative='two-sided')
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```

```python
### odd ratio smaller than one (sleeping less (run more) if drinking coffee)
t_stat, p_val = stats.fisher_exact(table, alternative='less') 
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```

```python
### odd ratio greater than one (sleeping more (run less) if drinking coffee)
t_stat, p_val = stats.fisher_exact(table, alternative='greater')
print("t-statistic = " + str(t_stat))  
print("p-value = " + str(p_val))
```
