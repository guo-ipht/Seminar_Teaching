# Part 2-Q2

```python
### define convex function
def f_x1(x, a=1):
    # b = a*2
    return a*(x**2 - 4*x + 10)

def df_x1(x, a=1):
    return a*(2*x - 4)

def ddf_x1(x, a=1):
    return a*2
```

```python
x = np.linspace(-7, 10, 100)
plt.plot(x, f_x1(x))
plt.xlabel('x')
plt.ylabel('f(x)')
```

```python
### define newton's method for optimization
def my_NM(x, f_x, df_x, ddf_x, x0, n_step=5, a=1):
    cur_x = x0
    for i in range(n_step):
        ### plot f(x)
        plt.plot(x, f_x(x, a))
        plt.xlabel('x')
        plt.ylabel('f(x)')
        
        ### plot f(cur_x)
        plt.scatter(cur_x, f_x(cur_x, a), c='red')
        plt.text(cur_x, f_x(cur_x, a), s=str(np.round(f_x(cur_x, a), 4)))
        
        ### update cur_x according to Newton's optimization step
        print(['cur_x', 'cur_df', 'cur_ddf'])
        print([np.round(cur_x, 4), np.round(df_x(cur_x, a), 4), np.round(ddf_x(cur_x, a), 4)])
        cur_x = cur_x - (df_x(cur_x, a)/(ddf_x(cur_x, a)))        
        
        plt.show()
    return cur_x, f_x(cur_x)
```

```python
### minimize f1(x)
x = np.linspace(-7, 10, 100)
x_opt, y_opt = my_NM(x, f_x1, df_x1, ddf_x1, x0=-3, a=1)
```

```python
### define non-convex function
def f_x2(x, a=1):
    return a*(x**5 - 4*x**4 - 2*x**3 + 3*x**2 + x + 10)

def df_x2(x, a=1):
    return a*(5*x**4 - 16*x**3 - 6*x**2 + 6*x + 1)

def ddf_x2(x, a=1):
    return a*(20*x**3 - 48*x**2 - 12*x + 6)
```

```python
x = np.array(range(-100, 101))/100.0
plt.plot(x, f_x2(x))
plt.xlabel('x')
plt.ylabel('f(x)')
```

```python
### minimize f2(x)
x = np.array(range(-100, 101))/100.0
my_NM(x, f_x2, df_x2, ddf_x2, x0=-0.95, a=1)
```

```python
### minimize f(x)
my_NM(x, f_x2, df_x2, ddf_x2, x0=0.8, a=-1)
```

```python
### maximize of f(x), means minimize -f(x)
my_NM(x, f_x2, df_x2, ddf_x2, x0=-0.3, a=1)
```

```python
### miminize of f(x)
x = np.array(range(-100, 101))/100.0

fun = lambda x: f_x2(x, a=1)
optim = spo.minimize(fun=fun, x0=np.array(-0.1), method='tnc', bounds=[(-1, 1)])

plt.plot(x, f_x2(x, a=1))
plt.xlabel('x')
plt.ylabel('f(x)')

### plot f(cur_x)
plt.scatter(optim['x'], optim['fun'], c='red')
plt.text(optim['x'], optim['fun'], s=str(np.round(optim['fun'], 4)))
```

```python
### to maximize f(x), we need miminize of -f(x)
fun = lambda x: f_x2(x, a=-1)
optim = spo.minimize(fun=fun, x0=np.array(-0.2), method='tnc', bounds=[(-1, 1)])

### plot f(x)
plt.plot(x, f_x2(x, a=1))
plt.xlabel('x')
plt.ylabel('f(x)')

### plot f(cur_x)
plt.scatter(optim['x'], -optim['fun'], c='red')
plt.text(optim['x'], -optim['fun'], s=str(np.round(-optim['fun'], 4)))
```

```python
import pandas as pd
```

```python
### load data and plot

data = pd.read_table('4aap_data.txt', sep=' ', header=None)
wn = data[0].values
spec = data[1].values

plt.plot(wn, spec)
```

```python
### find the most intensive peak
ix_max = np.argmax(spec)
peak_max = wn[ix_max]
peak_max
```

```python
### mark maximal peak:
plt.plot(wn, spec)
plt.plot(wn[(ix_max-7):(ix_max+7)], spec[(ix_max-7):(ix_max+7)])
```

```python
def gauss_func(x, O, A, b, c):
    """
    Parameters
    ----------
    x : numpy array
        input xaxis.
    O, A, b, c : int
        Initialization values.

    Returns
    -------
    numpy array
        Corresponding values using Gaussian function.
    """
    return O + A * np.exp(-(x - b) ** 2 / (2 * c ** 2))
   
def lorentz_func(x, O, A, b, c):
    """
    Parameters
    ----------
    x : numpy array
        input xaxis.
    O, A, b, c : int
        Initialization values.

    Returns
    -------
    numpy array
        Corresponding values using Lorentzian function.

    """
    return O + A * c / ((c ** 2) + (b - x) ** 2)
```

```python
x = wn[(ix_max-7):(ix_max+7)]
y = spec[(ix_max-7):(ix_max+7)]
O_0 = np.min(y)
b_0 = peak_max
c_0 = 5
A_0 = np.max(y) - np.min(y)

p0 = [O_0, A_0, b_0, c_0]  ### initial guess

### fit peak in to Gaussian:
popt, _ = spo.curve_fit(gauss_func, x, y, p0=p0, method='trf', bounds=(0, np.inf))
result = {'O': popt[0], 'A': popt[1], 'b': popt[2], 'c': popt[3]} 

fitted = gauss_func(x, result['O'], result['A'], result['b'], result['c'])

print('position: '+ str(np.round(result['b'],2)))
print('intensity: '+ str(np.round(np.max(fitted),2)))
print('width: '+str(np.round(2*result['c'],2)))

### plot fitting
plt.plot(x, y, label='original')
plt.plot(x, fitted, label='fitted')
plt.title('Gauss')
plt.legend()
```

```python
### fit peak in to Lorenz:
popt, _ = spo.curve_fit(lorentz_func, x, y, p0=p0, method='trf', bounds=(0, np.inf))
result = {'O': popt[0], 'A': popt[1], 'b': popt[2], 'c': popt[3]} 

fitted = lorentz_func(x, result['O'], result['A'], result['b'], result['c'])

print('position: '+ str(np.round(result['b'],2)))
print('intensity: '+ str(np.round(np.max(fitted),2)))
print('width: '+str(np.round(2*result['c'],2)))

### plot fitting
plt.plot(x, y, label='original')
plt.plot(x, fitted, label='fitted')
plt.title('Gauss')
plt.legend()
```
