```python
## define an array
out_dice = np.array((3, 2, 2, 4, 5, 1, 6, 1, 3, 4))

### take values smaller than 4
val_below_4 = out_dice[out_dice < 4]

### how many are smaller than 4 (absolute frequence)
absolute_freq_4 = len(val_below_4)

### how many trials in total
n_roll = len(out_dice)

### relative frequency
relative_freq_4 = absolute_freq_4/n_roll
relative_freq_4
```

# code for Q5

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
### define an array
vals = np.array((2.3, 1.7, 1.3, 2.7, 3.3, 1.0, 1.7, 2.0, 1.3, 4.0))

### arithmetic mean
val_mean = np.mean(vals)
print('mean: '+str(val_mean))

### empirical variance
val_var = np.var(vals)
print('var: '+str(val_var))
```

```python
### generate normal distribution based on calculated mean and std
vals_normal = np.random.normal(val_mean, np.sqrt(val_var), 5000)

### plot grades and normal distribution
plt.boxplot(vals, positions=[1])
plt.boxplot(vals_normal, positions=[2])
plt.xticks([1,2], labels=['grades', 'normal'])
```
