````python
### example task1: return maximal and minimal item from list [1, -2, 4, 9, 3, 5]

my_list = [1, -2, 4, 9, 3, 5]   ## define my list

current_max = my_list[0]   ## initial guess of the maximum
current_min = my_list[0]  ## initial guess of the minimum

for v in my_list:         ## for loop, the code under it will be executed a few times until going through all values in my_list
    
    if v>current_max:     ## if statement, the lines under it will be executed if condition C1 is fulfiled
        
        current_max = v   ## condition C1 fulfilled (a larger value appears), update my guess of the maximum

    elif v<current_min:   ## elif statement, the lines under it will be executed if condition C1 is NOT fulfiled but C2 is fulfilled
        
        current_min = v   ## condition C2 fulfilled (a smaller value appears), update my guess of the manimum
    
    else:                 ## else statement, the lines under it will be executed if both conditions C1 and C2 are NOT fulfiled
        
        continue          ## not do anything
    
print('Maximum is :' + str(current_max))  
print('Minimum is :' + str(current_min)) 
````

# Functions

* A function is a block of code which only runs when it is called.
* You can pass data, known as parameters, into a function.
* A function can return data as a result.

````python
def max_min(x):
    """
    This function finds the maximum and minimum items of an input list
    return two values, maximum and minimum, respectively
    """
    
    current_max = x[0]   
    current_min = x[0]   
    for v in x:         
        
        if v>current_max:     
            current_max = v   
    
        elif v<current_min:   
            current_min = v   

        else:                 
            continue          
        
    return current_max, current_min
````

````python
### example task1: return maximal and minimal item from list [1, -2, 4, 9, 3, 5]
max1, min1 = max_min( x = [1, -2, 4, 9, 3, 5] )
print('Maximum: ' + str(max1))  
print('Minimum: ' + str(min1)) 
````

````python
### example task1: return maximal and minimal item from list [10, -2, 3, 15, 4, 7, 9]
max2, min2 = max_min( x = [10, -2, 3, 15, 4, 7, 9] )
print('Maximum: ' + str(max2))  
print('Minimum: ' + str(min2))
````

# Modules
* A module can be considered as a code library.
* It is a file containing a set of functions you want to include in your application.
* The module name is same as the file (.py) name without '.py'
* To create a module just save the functions you want in a file with the file extension .py
* Use import MODULE_NAME before using the module
* The function within the module can be called via MODULE_NAME.FUN

````python
import module_exp

max3, min3 = module_exp.max_min( x = [10, -2, 3, 15, 4, 7, 9] )
print('Maximum: ' + str(max3))  
print('Minimum: ' + str(min3)) 
````

````python
### import function as it is
from module_exp import max_min 

max4, min4 = max_min( x = [10, -2, 3, 15, 4, 7, 9] )
print('Maximum: ' + str(max4))  
print('Minimum: ' + str(min4)) 
````

````python
### import the function with alias name
from module_exp import max_min as m

max5, min5 = m( x = [10, -2, 3, 15, 4, 7, 9] )
print('Maximum: ' + str(max5))  
print('Minimum: ' + str(min5)) 
````
